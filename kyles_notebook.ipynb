```python
def tonum(df):
    x=pd.DataFrame(index=df.index)
    x["price"]=df["price"].fillna(0)
    x["supported_languages"]=df["supported_languages"].apply(len)
    x["windows"]=df["windows"].astype(int)
    x["linux"]=df["linux"].astype(int)
    x["mac"]=df["mac"].astype(int)
    x["screenshots"]=df["screenshots"].apply(len)
    return x
```

```python
x_train=tonum(train_df)
x_val=tonum(val_df)
y_train=train_df["hit"]
y_val=val_df["hit"]
x_train.head()
```

```python
dummy=DummyClassifier(strategy="most_frequent")
dummy.fit(x_train,y_train)
pred=dummy.predict(x_val)
print("Acc: ", accuracy_score(y_val,pred))
print("Recall: ", recall_score(y_val,pred))
```

```python
model=LogisticRegression(max_iter=3000)
model.fit(x_train,y_train)
model_pred=model.predict(x_val)
print("Acc: ", accuracy_score(y_val,model_pred))
print("Recall: ", recall_score(y_val,model_pred))
```

```python
for col in x_train.columns:
    cols=[c for c in x_train.columns if c!=col]
    m=LogisticRegression(max_iter=3000)
    m.fit(x_train[cols],y_train)
    p=m.predict(x_val[cols])
    print("Removing: ", col)
    print("Acc: ", accuracy_score(y_val,p))
    print("Recall: ", recall_score(y_val,p))
```

```python
cols=["supported_languages","mac","screenshots"]
model2=LogisticRegression(max_iter=3000)
model2.fit(x_train[cols],y_train)
model2_pred=model2.predict(x_val[cols])
print("Acc: ", accuracy_score(y_val,model2_pred))
print("Recall: ", recall_score(y_val,model2_pred))
```

```python
x_train["full_audio_languages"]=train_df["full_audio_languages"].apply(len)
x_val["full_audio_languages"]=val_df["full_audio_languages"].apply(len)
```

```python
cols=["supported_languages","mac","screenshots","full_audio_languages"]
model3=LogisticRegression(max_iter=3000)
model3.fit(x_train[cols],y_train)
model3_pred=model3.predict(x_val[cols])
print("Acc: ", accuracy_score(y_val,model3_pred))
print("Recall: ", recall_score(y_val,model3_pred))
```

```python
for col in cols:
    rem = [c for c in cols if c != col]
    m = LogisticRegression(max_iter=3000)
    m.fit(x_train[rem],y_train)
    p = m.predict(x_val[rem])
    print("Removing: ", col)
    print("Acc: ", accuracy_score(y_val,p))
    print("Recall: ", recall_score(y_val,p))
```

```python
final_cols = ["supported_languages","mac","screenshots","full_audio_languages"]
final_model = LogisticRegression(max_iter=3000)
final_model.fit(x_train[final_cols],y_train)
final_pred = final_model.predict(x_val[final_cols])
print("Final Accuracy:", accuracy_score(y_val,final_pred))
print("Final Recall:", recall_score(y_val,final_pred))
```

```python
# The initial model used price, supported languages, windows, lunux, mac and screenshots. The effect on accuracy and recall by each feature was tested by removing each feature indivisually.
# Price, windows support and linux support had little impact on the results and were removed. Full audio languages was added to improve recall. Screenshots had the highest affect on
# recall. The final model has an accuracy of 74.06% with a recall of 3.84%.
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split
URL = (
    "https://huggingface.co/datasets/FronkonGames/steam-games-dataset/"
    "resolve/main/data/train-00000-of-00001.parquet"
)
df = pd.read_parquet(URL)
df["release_year"] = pd.to_datetime(df["release_date"], errors="coerce").dt.year
df = (
    df[df["release_year"].between(2006, 2024, inclusive="both")]
    .dropna(subset=["positive"])
    .copy()
)
df["hit"] = (
    df.groupby("release_year")["positive"]
    .transform(lambda s: s >= s.quantile(0.75))
    .astype(int)
)
train_val_df, test_df = train_test_split(
    df, test_size=0.20, random_state=42, stratify=df["hit"]
)
train_df, val_df = train_test_split(
    train_val_df, test_size=0.25, random_state=42, stratify=train_val_df["hit"]
)
print(f"Train: {len(train_df)} | Val: {len(val_df)} | Test: {len(test_df)}")
```
